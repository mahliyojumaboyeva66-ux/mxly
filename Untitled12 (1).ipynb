```python
!pip install -q importlib-resources

import gc
import time
import numpy as np
import pandas as pd
import tensorflow as tf
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
from scipy.ndimage import binary_erosion, binary_dilation
from tensorflow import keras
from tensorflow.keras import layers

SEED, IMG, BATCH, EPOCHS = 42, 128, 32, 8
AUTOTUNE = tf.data.AUTOTUNE
keras.utils.set_random_seed(SEED)

(train_raw, val_raw, test_raw), info = tfds.load(
    "oxford_iiit_pet",
    split=["train[:85%]", "train[85%:]", "test"],
    with_info=True)


def prep(d, train):
    img = tf.image.resize(tf.cast(d["image"], tf.float32), (IMG, IMG)) / 127.5 - 1.0
    m = tf.image.resize(tf.cast(d["segmentation_mask"], tf.float32), (IMG, IMG), method="nearest")
    m = tf.cast(m != 2, tf.float32)
    if train:
        z = tf.concat([img, m], axis=-1)
        z = tf.image.random_flip_left_right(z)
        img, m = z[..., :3], z[..., 3:]
    return img, m


train_ds = (train_raw.map(lambda d: prep(d, True), num_parallel_calls=AUTOTUNE)
            .shuffle(1000, seed=SEED).batch(BATCH).prefetch(AUTOTUNE))
val_ds = val_raw.map(lambda d: prep(d, False), num_parallel_calls=AUTOTUNE).batch(BATCH).prefetch(AUTOTUNE)
test_ds = test_raw.map(lambda d: prep(d, False), num_parallel_calls=AUTOTUNE).batch(BATCH).prefetch(AUTOTUNE)

test_imgs = np.concatenate([x.numpy() for x, _ in test_ds])
test_masks = np.concatenate([y.numpy() for _, y in test_ds])
print("Train/val/test:", info.splits["train"].num_examples, info.splits["test"].num_examples)


def conv_block(x, f):
    for _ in range(2):
        x = layers.Conv2D(f, 3, padding="same", use_bias=False)(x)
        x = layers.BatchNormalization()(x)
        x = layers.ReLU()(x)
    return x


def unet_scratch():
    inp = keras.Input(shape=(IMG, IMG, 3))
    x, skips = inp, []
    for f in [32, 64, 128, 256]:
        x = conv_block(x, f)
        skips.append(x)
        x = layers.MaxPooling2D()(x)
    x = conv_block(x, 512)
    for f, s in zip([256, 128, 64, 32], reversed(skips)):
        x = layers.Conv2DTranspose(f, 2, strides=2, padding="same")(x)
        x = layers.Concatenate()([x, s])
        x = conv_block(x, f)
    out = layers.Conv2D(1, 1, activation="sigmoid")(x)
    return keras.Model(inp, out)


def unet_pretrained():
    base = keras.applications.MobileNetV2(input_shape=(IMG, IMG, 3), include_top=False, weights="imagenet")
    names = ["block_1_expand_relu", "block_3_expand_relu", "block_6_expand_relu",
             "block_13_expand_relu", "block_16_project"]
    enc = keras.Model(base.input, [base.get_layer(n).output for n in names])
    inp = keras.Input(shape=(IMG, IMG, 3))
    feats = enc(inp)
    x = feats[-1]
    for f, s in zip([256, 128, 64, 32], reversed(feats[:-1])):
        x = layers.Conv2DTranspose(f, 3, strides=2, padding="same")(x)
        x = layers.Concatenate()([x, s])
        x = conv_block(x, f)
    x = layers.Conv2DTranspose(16, 3, strides=2, padding="same")(x)
    out = layers.Conv2D(1, 1, activation="sigmoid")(x)
    return keras.Model(inp, out)


def dice_loss(y_true, y_pred):
    y_true = tf.reshape(y_true, [tf.shape(y_true)[0], -1])
    y_pred = tf.reshape(y_pred, [tf.shape(y_pred)[0], -1])
    inter = tf.reduce_sum(y_true * y_pred, axis=1)
    denom = tf.reduce_sum(y_true, axis=1) + tf.reduce_sum(y_pred, axis=1)
    return 1.0 - tf.reduce_mean((2.0 * inter + 1.0) / (denom + 1.0))


LOSSES = {
    "BCE": keras.losses.BinaryCrossentropy(),
    "Dice": dice_loss,
    "Focal": keras.losses.BinaryFocalCrossentropy(gamma=2.0),
}
MODELS = {
    "U-Net (noldan)": (unet_scratch, 1e-3),
    "U-Net (MobileNetV2)": (unet_pretrained, 3e-4),
}


def boundary(m):
    return m ^ binary_erosion(m)


def boundary_f1(p, g, tol=2):
    bp, bg = boundary(p), boundary(g)
    if bp.sum() == 0 and bg.sum() == 0:
        return 1.0
    if bp.sum() == 0 or bg.sum() == 0:
        return 0.0
    prec = (bp & binary_dilation(bg, iterations=tol)).sum() / bp.sum()
    rec = (bg & binary_dilation(bp, iterations=tol)).sum() / bg.sum()
    return 0.0 if prec + rec == 0 else 2 * prec * rec / (prec + rec)


def score(prob, gt):
    p = prob[..., 0] > 0.5
    g = gt[..., 0] > 0.5
    inter = (p & g).sum((1, 2))
    union = (p | g).sum((1, 2))
    total = p.sum((1, 2)) + g.sum((1, 2))
    iou = np.where(union == 0, 1.0, inter / np.maximum(union, 1))
    dice = np.where(total == 0, 1.0, 2 * inter / np.maximum(total, 1))
    bf1 = np.array([boundary_f1(p[i], g[i]) for i in range(len(p))])
    return iou, dice, bf1


area = test_masks[..., 0].sum((1, 2))
small = area <= np.percentile(area, 10)

preds, rows = {}, []
for mname, (builder, lr) in MODELS.items():
    for lname, loss in LOSSES.items():
        keras.backend.clear_session()
        gc.collect()
        keras.utils.set_random_seed(SEED)
        model = builder()
        model.compile(optimizer=keras.optimizers.Adam(lr), loss=loss)
        t = time.perf_counter()
        model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS, verbose=0)
        train_time = time.perf_counter() - t
        prob = model.predict(test_ds, verbose=0)
        preds[(mname, lname)] = prob
        iou, dice, bf1 = score(prob, test_masks)
        rows.append({
            "Model": mname, "Loss": lname,
            "IoU": iou.mean(), "Dice": dice.mean(), "Boundary F1": bf1.mean(),
            "Dice (kichik obyekt)": dice[small].mean(),
            "Train vaqti (s)": train_time,
        })
        print(mname, lname, "tayyor")

res = pd.DataFrame(rows)
print(res.round(4).to_string(index=False))

show_model = "U-Net (MobileNetV2)"
worst_small = np.argsort(area)[:3]
rng = np.random.default_rng(SEED)
rand_ids = rng.choice(len(test_imgs), 3, replace=False)
ids = list(worst_small) + list(rand_ids)

fig, axes = plt.subplots(len(ids), 5, figsize=(14, 2.8 * len(ids)))
for r, i in enumerate(ids):
    axes[r, 0].imshow((test_imgs[i] + 1) / 2)
    axes[r, 0].set_title("Rasm" + (" (kichik obyekt)" if i in worst_small else ""), fontsize=9)
    axes[r, 1].imshow(test_masks[i, ..., 0], cmap="gray")
    axes[r, 1].set_title("Haqiqiy maska", fontsize=9)
    for c, lname in enumerate(LOSSES, start=2):
        axes[r, c].imshow(preds[(show_model, lname)][i, ..., 0] > 0.5, cmap="gray")
        axes[r, c].set_title(lname, fontsize=9)
    for c in range(5):
        axes[r, c].axis("off")
plt.tight_layout()
plt.show()
```
